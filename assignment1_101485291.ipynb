```python
# Author: Myles Romans Muirhead
# Assignment: #1

#b Create 4 variables:
gym_member = "Alex Alliton" #string
preferred_weight_kg = 20.5  #float
highest_reps = 25           #int
membership_active = True    #bool

#c Create a dictionary named "workout_stats" with the specified structure
#This is a dictionary with strings as keys and Tuple values that hold integers
workout_stats = {
    "Maddox": (20, 10, 50),
    "Nigel": (10, 20, 30),
    "Amir": (5, 15, 25),
}

#d Calculate total workout minutes for each friend using a loop
totals_to_add = {}
for friend, stats in workout_stats.items():
    total_minutes = sum(stats)
    totals_to_add[f"{friend}_Total"] = total_minutes

workout_stats.update(totals_to_add)

#e Create a 2-dimensional (nested) list called workout_list
# Workout_list is a list of lists the inner lists represent a "friend"'s workout minutes
workout_list = [list(stats) for stats in workout_stats.values() if isinstance(stats, tuple)]

#f Slice "workout_list" to parameters
yoga_running_minutes = [row[ :2] for row in workout_list]
print("Yoga and running minutes: ", yoga_running_minutes)

weightlift_minutes = [row[2] for row in workout_list[-2:]]
print("Weightlifting minutes: ", weightlift_minutes)

#g Check if any friend's total workout minutes are greater than or equal to 120

for friend, stats in workout_stats.items():
    if isinstance(stats, int) and stats >= 120:
        print(f"Great job staying active, {friend}!")

#h Allow user to input a friends name and check if the name exists in the dictionary
friend_name = input("Enter a friend's name: ")
if friend_name in workout_stats:
    print(f"{friend_name}'s workout minutes: ")
    print(f"Yoga: {workout_stats[friend_name][0]}")
    print(f"Running: {workout_stats[friend_name][1]}")
    print(f"Weightlifting: {workout_stats[friend_name][2]}")
    print(f"Total: {workout_stats[f'{friend_name}_Total']}")
else:
    print(f"Sorry, {friend_name} not found in the records.")

#i Include highest and lowest total workout minutes
totals = {friend: total for friend, total in workout_stats.items() if isinstance(total, int)}
max_friend = max(totals, key=totals.get)
min_friend = min(totals, key=totals.get)
print(f"Friend with the highest total workout minutes:{max_friend} {totals[max_friend]}")
print(f"Friend with the lowest total workout minutes: {min_friend} {totals[min_friend]}")
```
